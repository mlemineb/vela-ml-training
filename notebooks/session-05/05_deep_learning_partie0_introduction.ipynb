## 🧬 1. L'inspiration biologique

### Le neurone biologique
Un neurone biologique est composé de :
- **Dendrites** : reçoivent les signaux d'autres neurones
- **Corps cellulaire** : traite et intègre les signaux
- **Axone** : transmet le signal de sortie
- **Synapses** : connexions avec d'autres neurones

### Analogie avec l'informatique
- **Dendrites** → Entrées (inputs)
- **Corps cellulaire** → Fonction d'activation
- **Axone** → Sortie (output)
- **Synapses** → Poids (weights)

```python
def visualiser_neurone_biologique():
    """Visualisation simplifiée d'un neurone biologique"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Neurone biologique (schématique)
    ax1.set_xlim(0, 10)
    ax1.set_ylim(0, 6)
    
    # Dendrites
    for i in range(3):
        ax1.plot([0, 2], [2+i*0.8, 3], 'b-', linewidth=2, alpha=0.7)
        ax1.text(-0.5, 2+i*0.8, f'Signal {i+1}', fontsize=10, ha='right')
    
    # Corps cellulaire
    circle = plt.Circle((3, 3), 0.8, color='orange', alpha=0.7)
    ax1.add_patch(circle)
    ax1.text(3, 3, 'Corps\ncellulaire', ha='center', va='center', fontweight='bold')
    
    # Axone
    ax1.plot([3.8, 8], [3, 3], 'g-', linewidth=4)
    ax1.text(6, 3.5, 'Axone', fontsize=10, ha='center')
    
    # Sortie
    ax1.plot([8, 9.5], [3, 3], 'r-', linewidth=3)
    ax1.text(9.7, 3, 'Signal\nde sortie', fontsize=10, ha='left')
    
    ax1.set_title('🧬 Neurone Biologique', fontsize=14, fontweight='bold')
    ax1.axis('off')
    
    # Neurone artificiel
    ax2.set_xlim(0, 10)
    ax2.set_ylim(0, 6)
    
    # Entrées
    for i in range(3):
        ax2.plot([0, 2], [2+i*0.8, 3], 'b-', linewidth=2, alpha=0.7)
        ax2.text(-0.5, 2+i*0.8, f'x{i+1}', fontsize=12, ha='right', fontweight='bold')
        ax2.text(1, 2+i*0.8+0.2, f'w{i+1}', fontsize=10, ha='center', color='purple')
    
    # Fonction d'activation
    circle = plt.Circle((3, 3), 0.8, color='lightblue', alpha=0.7)
    ax2.add_patch(circle)
    ax2.text(3, 3, 'f(Σ)', ha='center', va='center', fontweight='bold', fontsize=14)
    
    # Sortie
    ax2.plot([3.8, 8], [3, 3], 'g-', linewidth=4)
    ax2.plot([8, 9.5], [3, 3], 'r-', linewidth=3)
    ax2.text(9.7, 3, 'y', fontsize=14, ha='left', fontweight='bold')
    
    ax2.set_title('🤖 Neurone Artificiel', fontsize=14, fontweight='bold')
    ax2.axis('off')
    
    plt.tight_layout()
    plt.show()

visualiser_neurone_biologique()
```

## ⚡ 2. Le neurone artificiel - Concepts fondamentaux

### Formule mathématique
Un neurone artificiel calcule :

**y = f(w₁x₁ + w₂x₂ + ... + wₙxₙ + b)**

Où :
- **x₁, x₂, ..., xₙ** : les entrées (inputs)
- **w₁, w₂, ..., wₙ** : les poids (weights) 
- **b** : le biais (bias)
- **f** : la fonction d'activation
- **y** : la sortie (output)

### Étapes de calcul
1. **Combinaison linéaire** : z = Σ(wᵢxᵢ) + b
2. **Activation** : y = f(z)

```python
def demo_neurone_calcul():
    """Démonstration du calcul dans un neurone"""
    print("🔢 Exemple de calcul dans un neurone artificiel")
    print("="*50)
    
    # Données d'exemple
    x = np.array([0.5, 0.8, 0.2])  # Entrées
    w = np.array([0.4, -0.6, 0.9])  # Poids
    b = 0.1  # Biais
    
    print(f"Entrées (x): {x}")
    print(f"Poids (w): {w}")
    print(f"Biais (b): {b}")
    print()
    
    # Étape 1: Combinaison linéaire
    z = np.dot(w, x) + b
    print(f"Étape 1 - Combinaison linéaire:")
    print(f"z = w₁x₁ + w₂x₂ + w₃x₃ + b")
    print(f"z = {w[0]}×{x[0]} + {w[1]}×{x[1]} + {w[2]}×{x[2]} + {b}")
    print(f"z = {z:.3f}")
    print()
    
    # Étape 2: Fonction d'activation (sigmoid)
    def sigmoid(z):
        return 1 / (1 + np.exp(-z))
    
    y = sigmoid(z)
    print(f"Étape 2 - Fonction d'activation (sigmoid):")
    print(f"y = σ(z) = 1/(1 + e^(-z))")
    print(f"y = 1/(1 + e^(-{z:.3f}))")
    print(f"y = {y:.3f}")
    
    return x, w, b, z, y

x, w, b, z, y = demo_neurone_calcul()
```

## 📊 3. Les fonctions d'activation

Les fonctions d'activation introduisent la **non-linéarité** dans le réseau.
Sans elles, même un réseau profond ne serait qu'une régression linéaire !

### Principales fonctions d'activation :

```python
def visualiser_fonctions_activation():
    """Visualise les principales fonctions d'activation"""
    x = np.linspace(-5, 5, 100)
    
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    axes = axes.flatten()
    
    # 1. Sigmoid
    sigmoid = 1 / (1 + np.exp(-x))
    axes[0].plot(x, sigmoid, 'b-', linewidth=2)
    axes[0].set_title('Sigmoid: σ(x) = 1/(1+e^(-x))')
    axes[0].grid(True, alpha=0.3)
    axes[0].set_ylim(-0.1, 1.1)
    
    # 2. Tanh
    tanh = np.tanh(x)
    axes[1].plot(x, tanh, 'g-', linewidth=2)
    axes[1].set_title('Tanh: tanh(x)')
    axes[1].grid(True, alpha=0.3)
    axes[1].set_ylim(-1.1, 1.1)
    
    # 3. ReLU
    relu = np.maximum(0, x)
    axes[2].plot(x, relu, 'r-', linewidth=2)
    axes[2].set_title('ReLU: max(0, x)')
    axes[2].grid(True, alpha=0.3)
    
    # 4. Leaky ReLU
    leaky_relu = np.where(x > 0, x, 0.1 * x)
    axes[3].plot(x, leaky_relu, 'orange', linewidth=2)
    axes[3].set_title('Leaky ReLU: max(0.1x, x)')
    axes[3].grid(True, alpha=0.3)
    
    # 5. Softmax (exemple avec 3 valeurs)
    x_soft = np.array([-1, 0, 1])
    softmax = np.exp(x_soft) / np.sum(np.exp(x_soft))
    axes[4].bar(['x₁=-1', 'x₂=0', 'x₃=1'], softmax, color=['red', 'green', 'blue'], alpha=0.7)
    axes[4].set_title('Softmax: e^(xᵢ)/Σe^(xⱼ)')
    axes[4].set_ylabel('Probabilité')
    
    # 6. Comparaison
    axes[5].plot(x, sigmoid, 'b-', label='Sigmoid', linewidth=2)
    axes[5].plot(x, tanh, 'g-', label='Tanh', linewidth=2)
    axes[5].plot(x, relu, 'r-', label='ReLU', linewidth=2)
    axes[5].set_title('Comparaison')
    axes[5].legend()
    axes[5].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Tableau récapitulatif
    print("📋 Récapitulatif des fonctions d'activation:")
    print("="*60)
    print("| Fonction  | Formule        | Avantages           | Inconvénients      |")
    print("|-----------|----------------|--------------------|--------------------|")
    print("| Sigmoid   | 1/(1+e^(-x))   | Sortie [0,1]       | Gradient qui disparaît |")
    print("| Tanh      | tanh(x)        | Sortie [-1,1]      | Gradient qui disparaît |")
    print("| ReLU      | max(0,x)       | Simple, efficace   | Neurones morts     |")
    print("| Leaky ReLU| max(0.1x,x)    | Pas de neurones morts | Hyperparamètre   |")
    print("| Softmax   | e^xi/Σe^xj     | Probabilités       | Seulement en sortie|")

visualiser_fonctions_activation()
```

## 🏗️ 4. Architecture des réseaux de neurones

### Types d'architectures :

1. **Perceptron simple** : 1 neurone, problèmes linéairement séparables
2. **Perceptron multi-couches (MLP)** : Plusieurs couches, problèmes non-linéaires
3. **Réseaux profonds** : Nombreuses couches cachées

### Terminologie importante :
- **Couche d'entrée** : reçoit les données
- **Couches cachées** : traitent l'information
- **Couche de sortie** : produit la prédiction
- **Poids et biais** : paramètres à apprendre

```python
def visualiser_architectures():
    """Visualise différentes architectures de réseaux"""
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    # 1. Perceptron simple
    ax = axes[0]
    ax.set_xlim(0, 4)
    ax.set_ylim(0, 4)
    
    # Entrées
    for i in range(3):
        ax.scatter(0.5, 1+i*0.8, s=100, c='lightblue', edgecolor='black')
        ax.text(0.2, 1+i*0.8, f'x{i+1}', fontsize=10, ha='center')
        ax.plot([0.5, 2.5], [1+i*0.8, 2], 'k-', alpha=0.5)
    
    # Neurone
    ax.scatter(2.5, 2, s=200, c='orange', edgecolor='black')
    ax.text(2.5, 2, 'y', fontsize=12, ha='center', fontweight='bold')
    
    ax.set_title('Perceptron Simple', fontweight='bold')
    ax.axis('off')
    
    # 2. MLP
    ax = axes[1]
    ax.set_xlim(0, 6)
    ax.set_ylim(0, 5)
    
    # Couche d'entrée
    for i in range(3):
        ax.scatter(0.5, 1.5+i*0.8, s=80, c='lightblue', edgecolor='black')
        ax.text(0.2, 1.5+i*0.8, f'x{i+1}', fontsize=9, ha='center')
    
    # Couche cachée
    for i in range(4):
        ax.scatter(2.5, 1+i*0.8, s=80, c='lightgreen', edgecolor='black')
        # Connexions entrée -> cachée
        for j in range(3):
            ax.plot([0.5, 2.5], [1.5+j*0.8, 1+i*0.8], 'k-', alpha=0.3, linewidth=0.5)
    
    # Couche de sortie
    for i in range(2):
        ax.scatter(4.5, 1.8+i*0.8, s=80, c='orange', edgecolor='black')
        ax.text(4.8, 1.8+i*0.8, f'y{i+1}', fontsize=9, ha='left')
        # Connexions cachée -> sortie
        for j in range(4):
            ax.plot([2.5, 4.5], [1+j*0.8, 1.8+i*0.8], 'k-', alpha=0.3, linewidth=0.5)
    
    ax.text(0.5, 0.5, 'Entrée', ha='center', fontweight='bold')
    ax.text(2.5, 0.5, 'Cachée', ha='center', fontweight='bold')
    ax.text(4.5, 0.5, 'Sortie', ha='center', fontweight='bold')
    ax.set_title('Perceptron Multi-Couches (MLP)', fontweight='bold')
    ax.axis('off')
    
    # 3. Réseau profond
    ax = axes[2]
    ax.set_xlim(0, 8)
    ax.set_ylim(0, 5)
    
    positions = [0.5, 2, 3.5, 5, 6.5]
    sizes = [3, 4, 3, 4, 2]
    colors = ['lightblue', 'lightgreen', 'lightgreen', 'lightgreen', 'orange']
    labels = ['Entrée', 'Cachée 1', 'Cachée 2', 'Cachée 3', 'Sortie']
    
    for layer, (pos, size, color, label) in enumerate(zip(positions, sizes, colors, labels)):
        for i in range(size):
            y_pos = 2.5 - (size-1)*0.3/2 + i*0.3
            ax.scatter(pos, y_pos, s=60, c=color, edgecolor='black')
            
            # Connexions
            if layer < len(positions)-1:
                next_size = sizes[layer+1]
                for j in range(next_size):
                    next_y = 2.5 - (next_size-1)*0.3/2 + j*0.3
                    ax.plot([pos, positions[layer+1]], [y_pos, next_y], 'k-', alpha=0.2, linewidth=0.3)
        
        ax.text(pos, 1, label, ha='center', fontweight='bold', fontsize=9)
    
    ax.set_title('Réseau Profond (Deep Network)', fontweight='bold')
    ax.axis('off')
    
    plt.tight_layout()
    plt.show()

visualiser_architectures()
```

## 🎯 5. Pourquoi le Deep Learning fonctionne-t-il ?

### Apprentissage de représentations hiérarchiques

Les réseaux profonds apprennent automatiquement des **représentations** de plus en plus abstraites :

1. **Couches basses** : caractéristiques simples (contours, textures)
2. **Couches moyennes** : motifs plus complexes (formes, objets partiels)  
3. **Couches hautes** : concepts abstraits (objets complets, scènes)

### Avantages du Deep Learning :
- **Apprentissage automatique des features** (pas besoin d'ingénierie manuelle)
- **Capacité à modéliser des relations complexes**
- **Performance sur des données non-structurées** (images, texte, audio)
- **Amélioration avec plus de données**

```python
def demo_apprentissage_hierarchique():
    """Démonstration conceptuelle de l'apprentissage hiérarchique"""
    fig, ax = plt.subplots(1, 1, figsize=(14, 8))
    
    # Simulation d'une image et de ses représentations
    np.random.seed(42)
    
    # Couche d'entrée (pixels)
    ax.text(1, 7, 'Image d\'entrée\n(pixels bruts)', ha='center', fontweight='bold', 
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue"))
    
    # Représentation visuelle des pixels
    pixels = np.random.rand(8, 8)
    im1 = ax.imshow(pixels, extent=[0.5, 1.5, 5.5, 6.5], cmap='gray', alpha=0.7)
    
    # Couche 1 - Détecteurs de contours
    ax.text(3, 7, 'Couche 1\n(contours, bordures)', ha='center', fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgreen"))
    
    # Simulation de filtres de contours
    for i, angle in enumerate([0, 45, 90, 135]):
        y_pos = 6.5 - i*0.3
        ax.plot([2.7, 3.3], [y_pos, y_pos], 'k-', linewidth=3, 
                transform=ax.transData, alpha=0.8)
        ax.text(3.5, y_pos, f'{angle}°', fontsize=8)
    
    # Couche 2 - Formes simples
    ax.text(5, 7, 'Couche 2\n(formes simples)', ha='center', fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightyellow"))
    
    # Simulation de formes
    shapes = ['○', '□', '△', '◇']
    for i, shape in enumerate(shapes):
        y_pos = 6.5 - i*0.3
        ax.text(5, y_pos, shape, fontsize=16, ha='center')
    
    # Couche 3 - Objets partiels
    ax.text(7, 7, 'Couche 3\n(parties d\'objets)', ha='center', fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral"))
    
    parts = ['👁️', '👃', '👂', '👄']
    for i, part in enumerate(parts):
        y_pos = 6.5 - i*0.3
        ax.text(7, y_pos, part, fontsize=14, ha='center')
    
    # Couche de sortie - Objets complets
    ax.text(9, 7, 'Sortie\n(objets complets)', ha='center', fontweight='bold',
            bbox=dict(boxstyle="round,pad=0.3", facecolor="orange"))
    
    objects = ['😊', '🐱', '🚗', '🏠']
    for i, obj in enumerate(objects):
        y_pos = 6.5 - i*0.3
        ax.text(9, y_pos, obj, fontsize=16, ha='center')
    
    # Flèches de connexion
    for x in [1.5, 3.5, 5.5, 7.5]:
        ax.arrow(x, 6.2, 1, 0, head_width=0.1, head_length=0.1, fc='black', ec='black')
    
    # Titre et labels
    ax.text(5, 8.5, 'Apprentissage Hiérarchique dans un Réseau Profond', 
            ha='center', fontsize=16, fontweight='bold')
    
    ax.text(5, 4.5, 'Abstraction croissante →', ha='center', fontsize=14, 
            style='italic', color='red')
    
    ax.set_xlim(0, 10)
    ax.set_ylim(4, 9)
    ax.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    print("🧠 Principe clé du Deep Learning:")
    print("Chaque couche apprend des représentations de plus en plus abstraites")
    print("des données d'entrée, permettant de résoudre des problèmes complexes.")

demo_apprentissage_hierarchique()
```
